```python
import pandas as pd
from ftplib import FTP
import os
import gzip
import shutil

# Load accessions
df = merged_df.copy()
os.makedirs("downloads", exist_ok=True)

# Convert GCA_001417885.1 → 001417885 and GCA_001417885.1 → GCA001417885v1
def numeric_accession(acc):
    return acc.split('_')[1].split('.')[0]

def formatted_accession(acc):
    base = numeric_accession(acc)
    version = acc.split('.')[-1]
    return f"GCA{base}v{version}"

ftp_host = "ftp.ensemblgenomes.ebi.ac.uk"
base_dir = "/pub/fungi/current/fasta"
numeric_map = {numeric_accession(a): a for a in df['assembly']}
formatted_map = {formatted_accession(a): a for a in df['assembly']}
fasta_map = {}

with FTP(ftp_host) as ftp:
    ftp.login()
    ftp.cwd(base_dir)
    top_level_dirs = ftp.nlst()

    for top_dir in top_level_dirs:
        full_top_path = f"{base_dir}/{top_dir}"

        if top_dir.startswith("fungi_"):
            # Dive into nested folders inside fungi_*
            try:
                ftp.cwd(full_top_path)
                inner_dirs = ftp.nlst()

                for inner_dir in inner_dirs:
                    dna_path = f"{full_top_path}/{inner_dir}/dna"
                    try:
                        ftp.cwd(dna_path)
                        files = ftp.nlst()

                        # Only check .dna.toplevel.fa.gz
                        for file in files:
                            if not file.endswith(".dna.toplevel.fa.gz"):
                                continue

                            for num_id, original_acc in numeric_map.items():
                                if num_id in file:
                                    gz_path = f"downloads/{file}"
                                    fasta_path = gz_path[:-3]

                                    if not os.path.exists(fasta_path):
                                        print(f"Downloading {file} from {dna_path}...")
                                        with open(gz_path, "wb") as out_file:
                                            ftp.retrbinary(f"RETR {file}", out_file.write)
                                        with gzip.open(gz_path, 'rb') as f_in, open(fasta_path, 'wb') as f_out:
                                            shutil.copyfileobj(f_in, f_out)

                                    fasta_map[original_acc] = os.path.basename(fasta_path)
                                    break  # Stop after one match
                        ftp.cwd(full_top_path)
                    except Exception as e:
                        print(f"Skipping {dna_path}: {e}")
            except Exception as e:
                print(f"Skipping {full_top_path}: {e}")
        else:
            # Normal folder: search in /dna for file with full formatted accession (GCAxxxxxxvY)
            dna_path = f"{full_top_path}/dna"
            try:
                ftp.cwd(dna_path)
                files = ftp.nlst()

                for file in files:
                    if not file.endswith(".dna.toplevel.fa.gz"):
                        continue

                    for full_id, original_acc in formatted_map.items():
                        if full_id in file:
                            gz_path = f"downloads/{file}"
                            fasta_path = gz_path[:-3]

                            if not os.path.exists(fasta_path):
                                print(f"Downloading {file} from {dna_path}...")
                                with open(gz_path, "wb") as out_file:
                                    ftp.retrbinary(f"RETR {file}", out_file.write)
                                with gzip.open(gz_path, 'rb') as f_in, open(fasta_path, 'wb') as f_out:
                                    shutil.copyfileobj(f_in, f_out)

                            fasta_map[original_acc] = os.path.basename(fasta_path)
                            break
                ftp.cwd(base_dir)
            except Exception as e:
                print(f"Skipping {dna_path}: {e}")

# Add column and save
df['fasta_filename'] = df['assembly'].map(fasta_map)
df.to_csv("accessions_with_fasta.csv", index=False)
print("\n✅ Done! Output saved to 'accessions_with_fasta.csv'")
```
